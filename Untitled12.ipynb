```python

# ==========================================
# Student Marks Management System
# ==========================================

SUBJECTS = ("Python", "SQL", "Computer Networks")
PASSING_MARKS = 35

# The student ID is the key; the record is the value.
students = {}


# 1. Get valid text input
def get_text(prompt):
    while True:
        value = input(prompt).strip()

        if value:
            return value

        print("Input cannot be empty. Try again.")


# 2. Get marks between 0 and 100
def get_marks(subject):
    while True:
        try:
            marks = int(input(f"Enter {subject} marks: "))

            if 0 <= marks <= 100:
                return marks

            print("Marks must be between 0 and 100.")

        except ValueError:
            print("Please enter a whole number, such as 85.")


# 3. Calculate student results
def calculate_result(marks):
    total = sum(marks)
    average = total / len(marks)

    # A student must pass every subject.
    passed = all(mark >= PASSING_MARKS for mark in marks)

    return {
        "total": total,
        "average": average,
        "passed": passed
    }


# 4. Add a student
def add_student(students):
    student_id = get_text("Enter student ID: ").upper()

    if student_id in students:
        print("This student ID already exists.")
        return

    name = get_text("Enter student name: ").title()

    marks = []

    for subject in SUBJECTS:
        mark = get_marks(subject)
        marks.append(mark)

    students[student_id] = {
        "name": name,
        "marks": marks
    }

    print(f"Student {name} added successfully!")


# 5. Display one student's result
def display_student(student_id, student):
    result = calculate_result(student["marks"])

    print("\n---------- STUDENT REPORT ----------")
    print("Student ID:", student_id)
    print("Name:", student["name"])

    for subject, mark in zip(SUBJECTS, student["marks"]):
        print(f"{subject}: {mark}")

    print("Total:", result["total"], "/", len(SUBJECTS) * 100)
    print("Average:", round(result["average"], 2))

    if result["passed"]:
        print("Result: PASS")
    else:
        print("Result: FAIL")

    print("------------------------------------")


# 6. Search using ID or name
def search_student(students):
    query = get_text("Enter student ID or name: ").strip().lower()

    for student_id, student in students.items():
        if (
            student_id.lower() == query
            or query in student["name"].lower()
        ):
            display_student(student_id, student)
            return

    # None means no matching student was found.
    student = None

    if student is None:
        print("Student not found.")


# 7. Display all students
def display_all_students(students):
    if not students:
        print("No students have been added yet.")
        return

    for student_id, student in students.items():
        display_student(student_id, student)


# 8. Main menu
def main():
    while True:
        print("\n===== STUDENT MANAGEMENT SYSTEM =====")
        print("1. Add student")
        print("2. Search student")
        print("3. Display all students")
        print("4. Exit")

        choice = input("Choose an option: ").strip()

        if choice == "1":
            add_student(students)

        elif choice == "2":
            search_student(students)

        elif choice == "3":
            display_all_students(students)

        elif choice == "4":
            print("Thank you for using the application!")
            break

        else:
            print("Invalid choice. Please select 1, 2, 3, or 4.")
            continue


# Run the application
if __name__ == "__main__":
    main()
```
